## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.

train_df.drop_duplicates(inplace=True)
test_df.drop_duplicates(inplace=True)
print("Duplicates removed successfully.")

for col in train_df.columns:
    if col not in ['Product ID', 'Type']:
        train_df[col] = pd.to_numeric(train_df[col], errors='coerce')


for col in test_df.columns:

    if col not in ['Product ID', 'Type']:
        test_df[col] = pd.to_numeric(test_df[col], errors='coerce')

num_cols = train_df.select_dtypes(include=np.number).columns.tolist()

cat_cols = train_df.select_dtypes(exclude=np.number).columns.tolist()
if 'Product ID' not in cat_cols and 'Product ID' in train_df.columns:
    cat_cols.append('Product ID')

for col in num_cols:
    if col in train_df.columns:
        if pd.api.types.is_numeric_dtype(train_df[col]):
            train_df[col].fillna(train_df[col].median(), inplace=True)
    if col in test_df.columns:
        if pd.api.types.is_numeric_dtype(test_df[col]):
            test_df[col].fillna(test_df[col].median(), inplace=True)

for col in cat_cols:
    if col in train_df.columns:
        train_df[col].fillna(train_df[col].mode()[0], inplace=True)
    if col in test_df.columns:
        test_df[col].fillna(test_df[col].mode()[0], inplace=True)

print("Missing values handled successfully.")

target_col = 'Machine failure' if 'Machine failure' in train_df.columns else None
features = [col for col in train_df.columns if col != target_col]

if 'id' in features:
    features.remove('id')

test_df = test_df[[col for col in features if col in test_df.columns]]

for col in cat_cols:
    if col in train_df.columns and col in test_df.columns:
        combined = pd.concat([train_df[col].astype(str), test_df[col].astype(str)], axis=0)
        le = LabelEncoder()
        le.fit(combined)
        train_df[col] = le.transform(train_df[col].astype(str))
        test_df[col] = le.transform(test_df[col].astype(str))

print("Label encoding completed safely.")

scaler = StandardScaler()
numeric_feature_cols = [col for col in features if col in train_df.select_dtypes(include=np.number).columns]

train_df[numeric_feature_cols] = scaler.fit_transform(train_df[numeric_feature_cols])

test_numeric_cols = [col for col in numeric_feature_cols if col in test_df.columns]
test_df[test_numeric_cols] = scaler.transform(
    test_df[test_numeric_cols]
)


print("Feature scaling applied successfully.")

if target_col:
    X = train_df[features]
    y = train_df[target_col]
else:
    X = train_df.copy()
    y = None

print("\n Final Dataset Ready:")
print("Train Features Shape:", X.shape)
print("Train Target Shape:", y.shape if y is not None else None)
print("Test Shape:", test_df.shape)
````

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Research Hypothesis

There is no significant difference in the average torque between machines that failed and machines that did not fail.

Null and Alternate Hypothesis

H₀ (Null Hypothesis):
Mean torque of failed machines = Mean torque of non-failed machines

H₁ (Alternate Hypothesis):
Mean torque of failed machines ≠ Mean torque of non-failed machines

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value
from scipy.stats import ttest_ind
failed = train_df[train_df['Machine failure'] == 1]['Torque [Nm]']
not_failed = train_df[train_df['Machine failure'] == 0]['Torque [Nm]']

t_stat, p_val = ttest_ind(failed, not_failed, equal_var=False)
print("T-Statistic:", t_stat)
print("P-Value:", p_val)
````

##### Which statistical test have you done to obtain P-Value?

I used the Independent Sample t-test.

##### Why did you choose the specific statistical test?

Because we are comparing mean values of a continuous variable (Torque) between two independent groups —
(machines that failed vs. machines that didn’t).

The t-test checks if the difference in means is statistically significant.

Conclusion Example:

If p-value < 0.05, we reject H₀ → meaning torque significantly differs between the two groups.
This suggests torque might influence machine failure.

## ***6. Feature Engineering & Data Pre-processing***

### 1. Handling Missing Values

````python
# Handling Missing Values & Missing Value Imputation

num_cols = train_df.select_dtypes(include=['int64', 'float64']).columns
cat_cols = train_df.select_dtypes(exclude=['int64', 'float64']).columns

for col in num_cols:
    if train_df[col].isnull().sum() > 0:
        train_df[col].fillna(train_df[col].median(), inplace=True)

for col in cat_cols:
    if train_df[col].isnull().sum() > 0:
        mode_val = train_df[col].mode()
        if len(mode_val) > 0:
            train_df[col].fillna(mode_val[0], inplace=True)
        else:
            train_df[col].fillna("Unknown", inplace=True)
````

#### What all missing value imputation techniques have you used and why did you use those techniques?

I used Median Imputation for numeric columns because the median is less affected by outliers compared to the mean,
making it a robust choice for skewed data.

I used Mode Imputation for categorical columns to fill missing entries with the most frequent category value.

This ensures no loss of data and keeps dataset consistency for model training.

After imputation, there were no missing values remaining in the dataset.

### 2. Handling Outliers

````python
# Handling Outliers & Outlier treatments

for col in ['Torque [Nm]', 'Rotational speed [rpm]', 'Tool wear [min]', 'Air temperature [K]', 'Process temperature [K]']:
    Q1 = train_df[col].quantile(0.25)
    Q3 = train_df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR
    train_df[col] = np.where(train_df[col] > upper_limit, upper_limit,
                             np.where(train_df[col] < lower_limit, lower_limit, train_df[col]))
````

##### What all outlier treatment techniques have you used and why did you use those techniques?

I used the IQR (Interquartile Range) method for outlier detection and capping.

Any value outside 1.5×IQR below Q1 or above Q3 was replaced (capped) with boundary values.

This method was chosen because it preserves data distribution while reducing the impact of extreme values.

It prevents outliers from biasing the model without deleting any data points.

### 3. Categorical Encoding

````python
# Encode your categorical columns

le = LabelEncoder()
cat_cols = ['Type']

for col in cat_cols:
    train_df[col] = le.fit_transform(train_df[col])
    test_df[col] = le.transform(test_df[col])
````

#### What all categorical encoding techniques have you used & why did you use those techniques?

I used this technique because:
