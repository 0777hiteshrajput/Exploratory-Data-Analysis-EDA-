## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
# 1. Handle Duplicates
titles = titles.drop_duplicates()
credits = credits.drop_duplicates()

# 2. Handle Missing Values
titles['runtime'] = titles['runtime'].fillna(titles['runtime'].median())
titles['imdb_score'] = titles['imdb_score'].fillna(titles['imdb_score'].mean())
titles['tmdb_score'] = titles['tmdb_score'].fillna(titles['tmdb_score'].mean())
titles['imdb_votes'] = titles['imdb_votes'].fillna(0)
titles['seasons'] = titles['seasons'].fillna(0)

# Fill text columns with 'Unknown'
text_columns = ['age_certification', 'genres', 'production_countries', 'description']
for col in text_columns:
    if col in titles.columns:
        titles[col] = titles[col].fillna('Unknown')

# 3. Correct Data Types
titles['release_year'] = titles['release_year'].astype(int)
titles['seasons'] = titles['seasons'].astype(int)
titles['runtime'] = titles['runtime'].astype(int)
titles['imdb_votes'] = titles['imdb_votes'].astype(int)

# 4. Clean Credits Dataset
print("Credits Dataset Columns:", credits.columns)

if 'character_name' in credits.columns:
    credits['character_name'] = credits['character_name'].fillna('Not Available')

if 'role' in credits.columns:
    credits['role'] = credits['role'].str.upper()

# 5. Merge datasets
df = pd.merge(titles, credits, on='id', how='left')

print("Dataset is now cleaned and ready!")
````
